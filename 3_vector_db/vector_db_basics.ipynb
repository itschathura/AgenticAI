### ChromaDB

```python
import chromadb
chroma_client = chromadb.Client()
```

```python
collection = chroma_client.create_collection(name="my_collection")
```

```python
collection.add(
    ids=["id1", "id2", "id3"],
    documents=[
        "This is a document about pineapple",
        "This is a document about oranges",
        "This is a document about smartphones and tvs"
    ]
)
```

```python
results = collection.query(
    query_texts=["This is a query related about phones"], # Chroma will embed this for you
    n_results=2 # how many results to return
)
print(results)
```

```python
result = collection.get(include=["documents", "metadatas", "embeddings"])

for id, document, metadata, embedding in zip(result["ids"], result["documents"], result["metadatas"], result["embeddings"]):
    print(id, document, metadata, embedding[:2])
    print(f"embedding length: {len(embedding)}")
```
